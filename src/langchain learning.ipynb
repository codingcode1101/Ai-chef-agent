```python
from dotenv import load_dotenv


load_dotenv()
```

```python
from langchain.tools import tool
def getWeather(location: str)-> str:
    """获取指定城市的当前天气信息。

    Args:
        location: 城市名称，例如 "Beijing"
    """
    return f"Current weather in {location} is sunny"
```

```python
from  langchain.agents import create_agent

agent = create_agent(
    model="deepseek-chat",
    tools = [getWeather]
)
```

```python
print("loading LLM...")
responses = agent.invoke({
    "messages" : [
        {"role": "user", "content": "What the weather like in Hangzhou?"}
    ]
})
```

```python
print(responses)
```

```python
for message in responses["messages"]:
    print(message.model_dump_json(indent=2))
    print()
```
